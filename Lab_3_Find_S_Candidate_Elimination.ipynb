## 2. Find-S Algorithm


```python
# Find-S starts with the most specific hypothesis and minimally generalizes it using positive examples.
def find_s(X,y):
  SH = ['phi'] * len(X[0])
  print(f"Initial Hypothesis: {SH}")
  for i in range(len(X)):
    if y[i].lower() == 'yes':
      for j in range(len(SH)):
        if SH[j].lower() == 'phi':
          SH[j] = X[i][j].lower()
        elif SH[j].lower() != X[i][j].lower():
          SH[j] = '?'
    print(f"SH on record {i+1}: {SH}")
  return SH

print(f"Final Hypothesis: {find_s(X,y)}")
```

## 3. Candidate Elimination Algorithm


```python
# Candidate Elimination maintains the specific (S) and general (G) boundaries of the version space.
def candidate_elimination(concepts,target):
  specific=concepts[0].copy()
  general=[["?" for i in range(len(specific))]]
  print("\nInitialization of specific and general hypothesis")
  print("\nSpecific Hypothesis:",specific)
  print("\nGeneral Hypothesis:",general)
  for i, h in enumerate(concepts):
    if target[i]=="Yes":
      for x in range(len(specific)):
        if specific[x]!=h[x]:
          specific[x]="?"
    else:
      for x in range(len(specific)):
        if h[x] != specific[x]:
          general.append(["?" if j != x else specific[x] for j in range(len(specific))])
    print(f"\nExample {i+1}")
    print("Specific Hypothesis:",specific)
    print("General Hypothesis:",general)
  return specific, general


S,G = candidate_elimination(X,y)
print("\nFinal Specific Hypothesis:",S)
print("\nFinal General Hypothesis:",list(k for k in G if not all(d=='?' for d in k)))
```
